## Atividade 1
Escreva os métodos regra de quadratura pelo ponto médio, Trapézio e Simpson dentro do arquivo $utils/algoritmos.py$.

````python
def medio(f, a, b):
    h = b - a
    return h * f((a + b) / 2)


def trapezio(f, a, b):
    h = b - a
    return h * (f(a) + f(b)) / 2


def simpson(f, a, b):
    h = (b - a) / 2
    return h * (f(a) + 4 * f((a + b) / 2) + f(b)) / 3


def integral(metodo, f, a, b, n=1e-3):
    if n <= 0:
        raise ValueError("n deve ser maior que zero.")

    s = 0.0
    c = a
    while c < b:
        d = min(c + n, b)
        s += metodo(f, c, d)
        c = d
    return s
````

````python
import sys
import os
import numpy as np
from scipy.integrate import quad

# Sobe um nível para a raiz do projeto e adiciona ao caminho do sistema
sys.path.append(os.path.abspath(os.path.join('..')))

from utils.algoritmos import (
    medio,
    trapezio,
    simpson,
    integral,
)


def main():
    a = 0
    b = 1

    funcoes = {
        "a) e^(-x)": lambda x: np.exp(-x),
        "b) x^2": lambda x: x**2,
        "c) x^3": lambda x: x**3,
        "d) x e^(-x^2)": lambda x: x * np.exp(-x**2),
        "e) 1/(x^2 + 1)": lambda x: 1 / (x**2 + 1),
        "f) x/(x^2 + 1)": lambda x: x / (x**2 + 1),
    }

    for nome, f in funcoes.items():
        print(nome)
        r_medio = integral(medio, f, a, b)
        r_trapezio = integral(trapezio, f, a, b)
        r_simpson = integral(simpson, f, a, b)
        r_scipy, _ = quad(lambda x: float(f(x)), a, b)
        print(f"Ponto medio = {r_medio:.8f}")
        print(f"Trapezio = {r_trapezio:.8f}")
        print(f"Simpson = {r_simpson:.8f}")
        print(f"SciPy = {r_scipy:.8f}")


if __name__ == "__main__":
    main()
````

## Atividade 2
Calcule numericamente o valor de  

$\displaystyle \int_{2}^{5} e^{4 - x^2} \, dx$  

usando os métodos **compostos** do **ponto médio**, **Trapézio** e **Simpson**.  
Obtenha os resultados utilizando, em cada quadratura, o número de pontos indicado.

**Respostas esperadas:**

| $n$ | Ponto médio | Trapézios | Simpson |
|:--:|:--:|:--:|:--:|
| 3 | 0.1056606 | 0.7503919 | 0.5005225 |
| 5 | 0.1726140 | 0.3964724 | 0.2784992 |
| 7 | 0.1973663 | 0.3062023 | 0.2393551 |
| 9 | 0.2084204 | 0.2721145 | 0.2306618 |

````python
f = lambda x: np.exp(4 - x**2)
a, b = 2.0, 5.0

def ponto_medio_composta(n):
    h = (b - a) / n
    return h * sum(f(a + (k + 0.5) * h) for k in range(n))

def trapezio_composto(n):
    h = (b - a) / (n - 1)
    pontos = np.linspace(a, b, n)
    return h * (0.5 * f(pontos[0]) + sum(f(pontos[1:-1])) + 0.5 * f(pontos[-1]))

def simpson_composto(n):
    if (n - 1) % 2 != 0:
        raise ValueError("Simpson composto exige um numero impar de pontos.")
    h = (b - a) / (n - 1)
    pontos = np.linspace(a, b, n)
    return h / 3 * (
        f(pontos[0])
        + f(pontos[-1])
        + 4 * sum(f(pontos[1:-1:2]))
        + 2 * sum(f(pontos[2:-1:2]))
    )

print("n | Ponto medio | Trapezios | Simpson")
for n in (3, 5, 7, 9):
    print(
        f"{n:2d} | {ponto_medio_composta(n):.7f} | "
        f"{trapezio_composto(n):.7f} | {simpson_composto(n):.7f}"
    )
````

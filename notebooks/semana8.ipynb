## Atividade 1
Para economizar energia elétrica, um agricultor implantou um sistema de painel solar na fazenda para alimentar uma bomba d’água, que faz a irrigação das plantações.  
A placa negra utilizada fica exposta ao sol e ao vento. Dessa forma, para gerar eletricidade, a irradiação solar $E$ (em $\text{W/m}^2$) sobre a placa tem que ser maior que a perda de calor por radiação e por condução, dada pela temperatura atmosférica $K$ (em Kelvin).

A equação é dada por:

$$
E = 5,67 \times 10^{-8} \, T^4 + 0,4(T - K)
$$

Determine a temperatura mínima da placa, dada a medição da irradiação e da temperatura do ar, sabendo que o valor médio diário é $E = 500,125$ e $K = 272,975$.

**Sugestão:** usar métodos da bissecção ou ponto fixo.  
**Resposta aproximada:**  

$$
T = 304{,}56801011987010952
$$

````python
# Define a função f(T)
def f(T):
    E = 500.125
    K = 272.975
    return 5.67e-8 * T**4 + 0.4 * (T - K) - E

# Função da sua biblioteca
def bissecao(f, a, b, TOL, iter):
    c = (a + b) / 2
    if f(a) * f(b) > 0:
        raise ValueError("Nenhuma raiz encontrada no intervalo.")
    else:
        i = 0
        ERRO = abs(f(b) - f(a))
        while ERRO > TOL and i < iter:
            c = (a + b) / 2.0
            if f(c) == 0:
                return c, i
            elif f(a) * f(c) < 0:
                b = c
            else:
                a = c
            i += 1
            ERRO = abs(f(b) - f(a))
        return c, i

# Define os parâmetros iniciais
a = 200.0          # Início do intervalo
b = 400.0          # Fim do intervalo
tolerancia = 1e-15 # Erro tolerado muito baixo para atingir a precisão da imagem
max_iter = 100     # Número máximo de iterações

# Executa o método
temperatura, iteracoes = bissecao(f, a, b, TOL=tolerancia, iter=max_iter)

# Imprime o resultado
print(f"Temperatura aproximada: T = {temperatura}")
````

## Atividade 2
Em um grande hospital, para evitar intermitência no sistema elétrico dos diversos equipamentos médicos, instalou-se três reatores de alta performance em série, com tensão nominal $R$ em função da potência máxima $P$ de operação dos enrolamentos de controle:

$$
17R_1 - 2R_2 - 3R_3 = P_1 = 500
$$  

$$
-5R_1 + 21R_2 - 2R_3 = P_2 = 200
$$  

$$
-5R_1 - 5R_2 + 22R_3 = P_3 = 300
$$  

Calcular as tensões nominais de cada reator.

**Sugestão:** usar o método de Gauss-Seidel.  
**Resposta aproximada:**  

$$
R_1 = 36{,}56081655798128
$$  

$$
R_2 = 20{,}768358378225123
$$  

$$
R_3 = 26{,}66572157641055
$$  


````python
import numpy as np

def seidel(A: np.ndarray, B: np.ndarray, k: int, TOL: float) -> np.ndarray:
    A = A.astype(float)
    B = B.astype(float)
    n = B.shape[0]
    X = np.zeros(n)

    for _ in range(k):
        Xk = X.copy()
        for i in range(n):
            s1 = np.dot(A[i, :i], X[:i])
            s2 = np.dot(A[i, i + 1 :], Xk[i + 1 :])
            X[i] = (B[i] - s1 - s2) / A[i, i]
        if np.linalg.norm(X - Xk, ord=2) < TOL:
            return X
    return X


# Definir a matriz de coeficientes A e o vetor de termos B
A = np.array([
    [17.0, -2.0, -3.0],
    [-5.0, 21.0, -2.0],
    [-5.0, -5.0, 22.0]
])

B = np.array([500.0, 200.0, 300.0])

# Resolução pelo Método de Gauss-Seidel
solucao_seidel = seidel(A, B, k=100, TOL=1e-14)

# Resolução exata pela Fatoração LU
#solucao_lu = lu(A, B)

print("Solução por Gauss-Seidel:")
print(f"R1 = {solucao_seidel[0]:.14f}")
print(f"R2 = {solucao_seidel[1]:.14f}")
print(f"R3 = {solucao_seidel[2]:.14f}")
````

## Atividade 3
Determine o valor da corrente no resistor **R3**, sabendo que $V_1 = V_2 = 100\,V$ e $R_1 = R_2 = R_3 = 10\,\Omega$.

### Circuito equivalente

            R2           R1
     ┌─────/\/\/\───┬──/\/\/\───┐
     │             │            │
     │             │            │
    V2(+)        /\/\/\      V1(+)
     │            R3            │
     │             │            │
     └─────────────┴────────────┘

$$
\begin{bmatrix}
20 & 10 \\
10 & 20
\end{bmatrix}
\begin{Bmatrix}
I_1 \\
I_2
\end{Bmatrix}
=
\begin{Bmatrix}
100 \\
100
\end{Bmatrix}
$$

**Sugestão:** usar o método de Gauss-Seidel.

**Resposta aproximada:**  

$$
I_{R3} = 6{,}6667\,A
$$

````python
import numpy as np

# Definir matriz de coeficientes e vetor de cargas
A = np.array([
    [20.0, 10.0],
    [10.0, 20.0]
])

B = np.array([100.0, 100.0])

# Resolução pelo método de Gauss-Seidel
I = seidel(A, B, k=100, TOL=1e-8)

I1, I2 = I[0], I[1]
I_R3 = I1 + I2

print(f"I1   = {I1:.4f} A")
print(f"I2   = {I2:.4f} A")
print(f"I_R3 = {I_R3:.4f} A")
````

## Atividade 4

Num sistema de energia solar, um vetor de equilíbrio energético $E$, na placa absorvente e na placa de vidro, segue o seguinte sistema de equações não lineares nas temperaturas absolutas $(K/m^2)$:

$$
(T_1^4 + 0,06823T_1) - (T_2^4 + 0,05848T_2) = E_1
$$    

$$
(T_1^4 + 0,05848T_1) - (2T_2^4 + 0,11696T_2) = E_2
$$  

Calcule o valor aproximado das temperaturas de equilíbrio, sabendo que os valores de equilíbrio energético foram medidos experimentalmente em:  

$E_1 = 0,01753$ e $E_2 = 0,00254$.

**Sugestão:** usar o método de Newton-Raphson.  
**Resposta aproximada:**  

$$
T_1 = 0,30543
$$  

$$ 
T_2 = 0,185261
$$ 

````python
import numpy as np


# Vetor de funções F(T)
def F(T):
    T1, T2 = T[0], T[1]
    f1 = (T1**4 + 0.06823 * T1) - (T2**4 + 0.05848 * T2) - 0.01753
    f2 = (T1**4 + 0.05848 * T1) - (2 * T2**4 + 0.11696 * T2) - 0.00254
    return np.array([f1, f2])


# Matriz Jacobiana J(T)
def Jacobiana(T):
    T1, T2 = T[0], T[1]
    j11 = 4 * T1**3 + 0.06823
    j12 = -4 * T2**3 - 0.05848
    j21 = 4 * T1**3 + 0.05848
    j22 = -8 * T2**3 - 0.11696
    return np.array([[j11, j12], [j21, j22]])


# Método de Newton-Raphson para sistemas
def newton_raphson_sistema(T0, TOL=1e-8, max_iter=100):
    T = np.array(T0, dtype=float)
    for k in range(max_iter):
        f_val = F(T)
        if np.linalg.norm(f_val, ord=2) < TOL:
            return T, k
        J_val = Jacobiana(T)
        delta_T = np.linalg.solve(J_val, -f_val)
        T = T + delta_T
    return T, max_iter


# Estimativa inicial (ex.: T1 = 0.5, T2 = 0.5)
T_inicial = [0.5, 0.5]
solucao, k = newton_raphson_sistema(T_inicial)

print(f"Convergência alcançada em {k} iterações:")
print(f"T1 = {solucao[0]:.6f}")
print(f"T2 = {solucao[1]:.6f}")
````
